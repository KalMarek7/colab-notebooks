```
# Start the server in the background
import subprocess
import threading
import time

def run_ollama_serve():
    subprocess.Popen(["ollama", "serve"])

thread = threading.Thread(target=run_ollama_serve)
thread.start()

# Wait a few seconds for the server to wake up
time.sleep(5)
print("🚀 Ollama server is up and running!")
```

```
import pandas as pd

# Load the CSV file
df_jobs = pd.read_csv('/content/jobs.csv')
display(df_jobs.head())
```

Now, let's iterate through the `text_content` column of the DataFrame and send each entry to the `test_extraction` function. We'll store the results in a new column.

```
if 'text_content' in df_jobs.columns:
    df_jobs['gemma3_4b'] = df_jobs['text_content'].apply(lambda x: test_extraction(x) if pd.notna(x) else None)
    print("Extraction complete. Displaying the first 5 rows with extracted data:")
    display(df_jobs[['text_content', 'gemma3_4b']].head())
else:
    print("Error: 'text_content' column not found in the CSV file.")
```

```
# Save the DataFrame with the new column to a CSV file
df_jobs.to_csv('results.csv', index=False)

print("DataFrame saved to 'results.csv'. You can now download it from the files section.")
```
